## 1.2.1 Linear Spaces

Linear spaces are sets of vectors that are closed under vector addition and scalar multiplication. In this section, I will demonstrate linear combinations, span, and linear independence using NumPy.

```python
import numpy as np
```

```python
u = np.array([1, 2])
v = np.array([3, 1])

print("u =", u)
print("v =", v)
```

### Linear Combination

Linear combinations are formed by multiplying vectors by scalar values, then adding the results. I will calculate 2u + 3v as an example.

```python
linear_combination = 2 * u + 3 * v

print("2u + 3v =", linear_combination)
```

The result is [11, 7] showing that a new vector can be formed by multiplying the original vectors by scalar values and adding them together.

### Span

The span of a set of vectors is the set of all vectors that can be created from their linear combinations. In this example, I will use different scalar values to create several vectors from u and v.

```python
combinations = [
    1 * u + 1 * v,
    2 * u + 1 * v,
    1 * u + 2 * v,
    -1 * u + 2 * v
]

for i, result in enumerate(combinations, 1):
    print(f"Combination {i}: {result}")
```

Each result above is a different linear combination of u and v. Since the scalar values can be changed, many different vectors can be made. The collection of all vectors that can be made this way is the span of u and v.

### Linear Independence

A set of vectors is linearly independent if none of the vectors can be written as a linear combination of the others. For vectors u and v, I will check whether one vector is a scalar multiple of the other.

```python
# Check whether v is a scalar multiple of u
ratio = v[0] / u[0]

print("Ratio between corresponding first components:", ratio)
print("Ratio between corresponding second components:", v[1] / u[1])

if np.isclose(v[1] / u[1], ratio):
    print("u and v are linearly dependent.")
else:
    print("u and v are linearly independent.")
```

The ratios of the corresponding components are different, so v cannot be written as a scalar multiple of u. This means u and v are linearly independent. This means that neither vector can be produced by scaling the other.

### Basis and Dimension

Because u and v are linearly independent and are vectors in R², they form a basis for R². So, the dimension of the space they span is 2.

```python
A = np.column_stack((u, v))

print("Matrix formed from u and v:")
print(A)
print("Rank:", np.linalg.matrix_rank(A))
```

## 1.2.2 Orthogonality

Orthogonality is when two vectors are perpendicular. Two vectors are orthogonal when their dot product is zero. I will show the dot product, vector norms, and projection using NumPy.

```python
a = np.array([1, 2])
b = np.array([2, -1])

dot_product = np.dot(a, b)

print("a =", a)
print("b =", b)
print("Dot product:", dot_product)
```

The dot product of a and b is 0. The two vectors are orthogonal, which means they are perpendicular to each other.

### Vector Norms

The norm of a vector shows its length. For a vector x, the Euclidean norm is shown as

$$
\|x\| = \sqrt{x_1^2 + x_2^2 + \cdots + x_n^2}
$$

I will find the norms of a and b.

```python
norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)

print("||a|| =", norm_a)
print("||b|| =", norm_b)
```

Both vectors have the same norm, about 2.236. The norm measures the length or magnitude of a vector.

### Projection

The projection of one vector onto another gives the component of one vector that lies in the direction of the other. The projection of a onto b is

$$
\operatorname{proj}_b(a) =
\frac{a \cdot b}{b \cdot b}b
$$

I will calculate this projection using NumPy.

```python
projection = (np.dot(a, b) / np.dot(b, b)) * b

print("Projection of a onto b:", projection)
```

The projection is the zero vector because a and b are orthogonal. Since a has no component in the direction of b, its projection onto b is 0.

### Checking for Orthogonality

Two vectors are orthogonal when their dot product is 0. I will check the dot product above.

```python
if np.isclose(np.dot(a, b), 0):
    print("The vectors are orthogonal.")
else:
    print("The vectors are not orthogonal.")
```

The dot product is 0, confirming that a and b are orthogonal. The projection of a onto b is also the zero vector, which is consistent with the fact that a has no component in the direction of b.

## 1.2.3 Gram-Schmidt Process

The Gram-Schmidt Process converts a set of linearly independent vectors to an orthonormal set of vectors that spans the same space. I will apply this process to three vectors in R3.

```python
v1 = np.array([1, 1, 0], dtype=float)
v2 = np.array([1, 0, 1], dtype=float)
v3 = np.array([0, 1, 1], dtype=float)

print("v1 =", v1)
print("v2 =", v2)
print("v3 =", v3)
```

### Step 1: First Orthogonal Vector

The first orthogonal vector is the first vector. I will call it u1.

```python
u1 = v1

print("u1 =", u1)
```

### Step 2: Second Orthogonal Vector

To make the second vector orthogonal to the first, I subtract the projection of v2 onto u1 from v2.

```python
projection_v2 = (np.dot(v2, u1) / np.dot(u1, u1)) * u1
u2 = v2 - projection_v2

print("Projection of v2 onto u1:", projection_v2)
print("u2 =", u2)
```

### Step 3: Third Orthogonal Vector

For the third vector, I subtract its projections onto both u1 and u2. This makes the resulting vector orthogonal to both previous vectors.

```python
projection_v3_u1 = (np.dot(v3, u1) / np.dot(u1, u1)) * u1
projection_v3_u2 = (np.dot(v3, u2) / np.dot(u2, u2)) * u2

u3 = v3 - projection_v3_u1 - projection_v3_u2

print("Projection onto u1:", projection_v3_u1)
print("Projection onto u2:", projection_v3_u2)
print("u3 =", u3)
```

### Step 4: Normalize the Orthogonal Vectors

To create an orthonormal set, I must divide each orthogonal vector by its norm. This gives each vector a length of 1 while keeping its direction.

```python
q1 = u1 / np.linalg.norm(u1)
q2 = u2 / np.linalg.norm(u2)
q3 = u3 / np.linalg.norm(u3)

print("q1 =", q1)
print("q2 =", q2)
print("q3 =", q3)
```

### Checking the Orthonormal Basis

The vectors should each have a norm of 1 and should be orthogonal. I will check to make sure this is true.

```python
print("Norms:")
print(np.linalg.norm(q1))
print(np.linalg.norm(q2))
print(np.linalg.norm(q3))

print("\nDot products:")
print("q1 · q2 =", np.dot(q1, q2))
print("q1 · q3 =", np.dot(q1, q3))
print("q2 · q3 =", np.dot(q2, q3))
```

The three vectors all have a norm of 1, and their dot products are 0. Therefore, the Gram-Schmidt Process did make an orthonormal set of vectors.

```python
Q = np.column_stack((q1, q2, q3))

print("Q =")
print(Q)

print("\nQ.T @ Q =")
print(Q.T @ Q)
```

The matrix Q has the three orthonormal vectors as columns. Since Q.T @ Q is about equal to the identity matrix, the vectors are orthonormal.

## 1.2.4 Eigenvalues and Eigenvectors

For a square matrix A, an eigenvector is a nonzero vector which direction is not changed when multiplied by A. The multiplication only scales the vector. This relationship is written as A v = lambda v, where lambda is the eigenvalue.

```python
A = np.array([
    [3, 1],
    [1, 3]
], dtype=float)

print("Matrix A:")
print(A)
```

### Calculating Eigenvalues and Eigenvectors

I will use NumPy to calculate the eigenvalues and corresponding eigenvectors of A.

```python
eigenvalues, eigenvectors = np.linalg.eig(A)

print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```

### Checking the Eigenpairs

For each eigenvalue and eigenvector, I will check that A v = lambda v. The residual A v - lambda v should be close to zero.

```python
for i, eigenvalue in enumerate(eigenvalues):
    vector = eigenvectors[:, i]

    Av = A @ vector
    lambda_v = eigenvalue * vector
    residual = Av - lambda_v

    print(f"Eigenpair {i + 1}")
    print("Eigenvalue:", eigenvalue)
    print("Eigenvector:", vector)
    print("A @ v:", Av)
    print("lambda * v:", lambda_v)
    print("Residual norm:", np.linalg.norm(residual))
    print()
```

The residual norms are 0, showing that the calculated eigenvalues and eigenvectors do satisfy A v = lambda v. The eigenvectors show directions that are retained by the matrix, while the eigenvalues show how those directions are scaled.

### Orthogonality of the Eigenvectors

Since A is symmetric, its eigenvectors should be orthogonal. I will check this using their dot product.

```python
v1 = eigenvectors[:, 0]
v2 = eigenvectors[:, 1]

print("Dot product of the eigenvectors:", np.dot(v1, v2))
```

The dot product of the two eigenvectors is 0, which means they are orthogonal. This is consistent with the property that eigenvectors that correspond to certain: eigenvalues of a symmetric matrix are orthogonal.
