## 2.3.1 Joint Probability Distributions

A joint probability distribution describes the probabilities of two random variables occurring together. Each combination of possible values has a probability, and all of the probabilities in the distribution should add up to 1.

I will use two simple random variables to represent the outcome of two coin flips. The first variable represents the result of the first flip and the second variable represents the result of the second flip.

```python
# Joint outcomes for two fair coin flips
outcomes = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

probabilities = np.array([
    0.25,
    0.25,
    0.25,
    0.25
])

print("Possible outcomes:")
print(outcomes)

print("\nJoint probabilities:")
print(probabilities)

print("\nSum of all joint probabilities:", probabilities.sum())
```

The four possible combinations of the two coin flips are equally likely, so each combination has a probability of 0.25. The probabilities add up to 1, which confirms that the joint distribution accounts for all possible outcomes.

### Visualizing the Joint Distribution

A small matrix makes it easier to see the probability associated with each combination of the two random variables. Each row represents the result of the first flip, while each column represents the result of the second flip.

```python
joint_distribution = np.array([
    [0.25, 0.25],
    [0.25, 0.25]
])

print("Joint probability distribution:")
print(joint_distribution)

print("\nRow sums:")
print(joint_distribution.sum(axis=1))

print("\nColumn sums:")
print(joint_distribution.sum(axis=0))

plt.imshow(joint_distribution)

plt.colorbar(label="Probability")
plt.xlabel("Second Flip")
plt.ylabel("First Flip")
plt.title("Joint Probability Distribution")

plt.xticks([0, 1], ["0", "1"])
plt.yticks([0, 1], ["0", "1"])

plt.show()
```

The matrix shows that each combination of the two coin flips has the same probability. The row and column sums give the probabilities for the individual random variables. The heatmap also makes it easy to see that all four combinations have equal probability.

## 2.3.2 Correlation and Dependence

Correlation measures the strength and direction of a linear relationship between two variables. A correlation close to 1 means the variables tend to increase together, while a correlation close to -1 means one tends to increase as the other decreases.

Dependence is a broader idea. Two variables can be dependent even when their correlation is not strong because correlation mainly measures linear relationships.

I will use a small example comparing hours spent studying with test scores.

```python
study_hours = np.array([1, 2, 3, 4, 5, 6])
test_scores = np.array([62, 67, 71, 78, 84, 89])

correlation = np.corrcoef(study_hours, test_scores)[0, 1]

print("Study hours:", study_hours)
print("Test scores:", test_scores)
print("Correlation:", correlation)
```

The correlation is close to 1, showing a strong positive linear relationship between study hours and test scores in this example. As study hours increase, the test scores also tend to increase.

### Visualizing Correlation

A scatter plot helps show the relationship between the two variables. If the points follow an upward pattern, that supports the positive correlation shown by the numerical calculation.

```python
plt.scatter(study_hours, test_scores)

plt.xlabel("Study Hours")
plt.ylabel("Test Score")
plt.title("Study Hours and Test Scores")

plt.show()
```

The scatter plot shows an upward pattern, which agrees with the high positive correlation. The points are fairly close to a straight-line pattern, so study hours and test scores have a strong positive linear relationship in this example.

## 2.3.3 Random Samples

A random sample is a group of observations selected from a larger population. Random sampling gives each member of the population a chance to be selected and can be used to study a population without collecting data from every individual.

I will create a simple population of test scores and randomly select a sample from it. I will use a fixed random seed so that the example gives the same result each time the notebook is run.

```python
np.random.seed(42)

population = np.array([
    58, 62, 65, 67, 70,
    72, 74, 76, 78, 80,
    82, 84, 86, 88, 91,
    93, 95, 97, 99
])

sample = np.random.choice(population, size=6, replace=False)

print("Population:")
print(population)

print("\nRandom sample:")
print(sample)

print("\nPopulation mean:", np.mean(population))
print("Sample mean:", np.mean(sample))
```

The random sample contains six observations selected from the larger population. The sample mean is not necessarily equal to the population mean, but it gives an estimate of the population mean. This shows how a smaller random sample can be used to learn about a larger population.

### Comparing the Population and Sample

A histogram can show how the sample compares with the overall population. Because the sample is much smaller, its distribution will not necessarily match the population perfectly.

```python
plt.hist(population, bins=6, alpha=0.6, label="Population")
plt.hist(sample, bins=6, alpha=0.6, label="Sample")

plt.xlabel("Test Score")
plt.ylabel("Frequency")
plt.title("Population and Random Sample")
plt.legend()

plt.show()
```

The histogram shows that the sample comes from the same population but contains fewer observations. Because random samples contain only part of the population, the sample distribution can look different from the full population. With a larger random sample, the sample would generally give a more representative picture of the population.
